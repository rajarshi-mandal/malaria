```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
ast = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Average Surface Temperature.csv")
ast.columns.values[-1] = "Average Surface Temperature"
csw = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Canopy Surface Water.csv")
csw.columns.values[-1] = "Canopy Surface Water"
et = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Evapotranspiration.csv")
et.columns.values[-1] = "Evapotranspiration"
nsat = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Near Surface Air Temperature.csv")
nsat.columns.values[-1] = "Near Surface Air Temperature"
pp = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Precipitation.csv")
pp.columns.values[-1] = "Precipitation"
sm = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Soil Moisture (10 to 40 cm underground).csv")
sm.columns.values[-1] = "Soil Moisture (10 to 40 cm underground)"
srhd = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Surface Relative Humidity (Daytime).csv")
srhd.columns.values[-1] = "Surface Relative Humidity (Daytime)"
srhn = pd.read_csv("/kaggle/input/earth-data-giovanni/raw/Surface Relative Humidity (Nighttime).csv")
srhn.columns.values[-1] = "Surface Relative Humidity (Nighttime)"
```

```python
merged = ast
datasets = [csw, et, nsat, pp, sm, srhd, srhn]
for df in datasets:
    merged = pd.merge(merged, df, on=["lat", "lon"], how="outer")
merged.to_csv('combined_data.csv', index=False)
```
